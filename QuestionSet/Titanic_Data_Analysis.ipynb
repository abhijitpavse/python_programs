# MODULE 6 — Age Feature Engineering

```python
# Create AgeGroup
age_bins = [
    -np.inf,
    12,
    17,
    30,
    50,
    65,
    np.inf
]

age_labels = [
    'Child',
    'Teenager',
    'Young Adult',
    'Adult',
    'Senior',
    'Elderly'
]

df['AgeGroup'] = pd.cut(
    df['AgeFilled'],
    bins=age_bins,
    labels=age_labels
)
```

```python
# IsChild

# The assignment specifically asks for NumPy conditional logic.

df['IsChild'] = np.where(
    df['AgeFilled'] < 18,
    1,
    0
)
```

# MODULE 13 — NumPy Feature Engineering

```python
# Fare Above Median
fare_median = df['Fare'].median()

df['FareAboveMedian'] = np.where(
    df['Fare'] > fare_median,
    1,
    0
)
```

```python
# FareLevel using np.select()
q1 = df['Fare'].quantile(0.25)
q2 = df['Fare'].quantile(0.50)
q3 = df['Fare'].quantile(0.75)

conditions = [
    df['Fare'] <= q1,
    (df['Fare'] > q1) & (df['Fare'] <= q2),
    (df['Fare'] > q2) & (df['Fare'] <= q3),
    df['Fare'] > q3
]

choices = [
    'Low',
    'Medium',
    'High',
    'Premium'
]

df['FareLevel'] = np.select(
    conditions,
    choices,
    default='Unknown'
)
```

```python
AgeNumericGroup
age_conditions = [
    df['AgeFilled'] <= 12,
    (df['AgeFilled'] > 12) & (df['AgeFilled'] <= 17),
    (df['AgeFilled'] > 17) & (df['AgeFilled'] <= 30),
    (df['AgeFilled'] > 30) & (df['AgeFilled'] <= 50),
    (df['AgeFilled'] > 50) & (df['AgeFilled'] <= 65),
    df['AgeFilled'] > 65
]

age_choices = [1, 2, 3, 4, 5, 6]

df['AgeNumericGroup'] = np.select(
    age_conditions,
    age_choices,
    default=0
)
```

```python
# LargeFamily
df['LargeFamily'] = np.where(
    df['FamilySize'] > 4,
    1,
    0
)
```
