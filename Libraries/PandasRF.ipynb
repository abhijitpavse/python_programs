```python
students = {
    "Student_ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "Name": [
        "Rahul", "Amit", "Priya", "Neha",
        "Arjun", "Sneha", "Rohan", "Kavya"
    ],
    "Age": [21, 22, 20, 21, 23, 20, 22, 21],
    "Gender": [
        "Male", "Male", "Female", "Female",
        "Male", "Female", "Male", "Female"
    ],
    "City": [
        "Mumbai", "Pune", "Mumbai", "Delhi",
        "Pune", "Mumbai", "Delhi", "Nashik"
    ],
    "Marks": [85, 72, 92, 64, 76, 95, 58, 88],
    "Attendance": [90, 78, 96, 82, 75, 98, 68, 91]
}

df = pd.DataFrame(students)

print(df)
```

```python
# print data with marks more than 80
df[df["Marks"] > 80] # same will be aplied for other conditions
```

```python
# marks are greater than 80 and attendance is greater than 95 
df[(df["Marks"] > 80) & (df["Attendance"] > 95)]
```

```python
# city is mumbai and pune
print(df[(df["City"] == "Mumbai") | (df["City"] == "Pune")])
```

```python
# city is not mumbai
print(df[~(df["City"] == "Mumbai")])
```

```python
# names where marks are above 80
df[df["Marks"] > 80]["Name"]

#OR 

print(df.loc[df["Marks"] > 80, "Name"])
```

```python
# OR(|) using isin
print(df[df["City"].isin(["Mumbai","Pune"])])
```
