```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y, color="Orange",linewidth=2,linestyle="--",marker="*",markerfacecolor="red",markeredgecolor="red")
plt.title("Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

# Pie Chart

```python
# import matplotlib.pyplot as plt
import numpy as np

y = np.array([35, 25, 25, 15])
mylabels = ["Apples", "Bananas", "Cherries", "Dates"]
myexplode = [0.2, 0, 0, 0]  # explode 1st slice

plt.pie(y, labels=mylabels, explode=myexplode, shadow=True, autopct='%1.1f%%')
plt.show()
```
