```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y = [10, 20, 30, 40, 50]
plt.plot(x, y)
plt.title("Sales report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show
```

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y, color="Orange",linewidth=2,linestyle="--",marker="*",markerfacecolor="red",markeredgecolor="red")
plt.title("Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y, color="Orange",linewidth=2,linestyle="--",marker="*",markerfacecolor="red",markeredgecolor="red")
plt.title("Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y, color="Orange",linewidth=2,linestyle="--",marker="o")
plt.title("Sales Report")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.grid()
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,10,100)
y = np.sin(x)
plt.plot(x,y)
plt.show()
# print(y)
```

```python
x = np.linspace(0,10,100)
y = np.cos(x)
plt.plot(x,y)
plt.show()
```

```python
x = np.linspace(0,10,100)
y = np.tan(x)
plt.plot(x,y)
plt.show()
```

```python
categories = ["A", "B", "C", "D"]
sales = [100, 150, 120, 180]
plt.bar(categories, sales)
plt.show()
```

```python
categories = ["A", "B", "C", "D"]
sales = [100, 150, 120, 180]
plt.barh(categories, sales)
plt.grid()
plt.show()
```
