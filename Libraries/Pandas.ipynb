# Create Labels
**With the index argument, you can name your own labels.**

```python
a = [1, 7, 2]

myvar = pd.Series(a, index = ["x", "y", "z"])

print(myvar["x"])
```

**Key/Value Objects as Series**
- Create a simple Pandas Series from a dictionary:

```python
calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories)

print(myvar)
```

```python
# Create a Series using only data from "day1" and "day2":

calories = {"day1": 420, "day2": 380, "day3": 390}

myvar = pd.Series(calories, index = ["day1", "day2"])

print(myvar)
```

# Pandas DataFrames

A Pandas DataFrame is a 2 dimensional data structure, like a 2 dimensional array, or a table with rows and columns.

```python
data = {
  "calories": [420, 380, 390],
  "duration": [50, 40, 45]
}

#load data into a DataFrame object:
df = pd.DataFrame(data)

print(df)
```

```python
# Locate Row

#refer to the row index:
print(df.loc[0])
```

```python
print(df.loc[[0, 1]])
```

# iloc 
integer location

```python
marks = pd.Series([85,90,78],
    index = ['Rahul', 'Abhijit', 'Rohit'],
    name = 'Student Marks')

print(marks)
print(marks['Abhijit'])
print(marks.iloc[1])  # using iloc

print(marks[marks > 90])

print(marks.head(1))
print(marks.tail(1))

print("sum:",marks.sum())
print("mean:",marks.mean())
print("min:",marks.min())
print("max",marks.max())
print("median:",marks.median())
print("count:",marks.count())
```

```python
cities = pd.Series([
    "Mumbai",
    "Pune",
    "Mumbai",
    "Delhi",
    "Pune",
    "Mumbai",
])
# if we write 'mumbai' And 'Mumbai' will be shown differently bcz its case sensitive
print(cities.value_counts())
```

```python
# for none value

cities = pd.Series([
    "Mumbai",
    "Pune",
    "Mumbai",
    None,
    "Delhi",
    "Pune",
    None,
    None,
    "Mumbai",
])
print(cities.value_counts())
print(cities.isnull())
print(cities.isnull().sum())

print(cities.notnull())
print(cities.notnull().sum)

# print(cities.dropna())  # to drop null value

c = cities.fillna("Entertainment")
print(c)
```

```python
# same as above

cities = pd.Series([
  "mumbai",
  "pune",
  "Mumbai",
  None,
  "delhi",
  "pune",
  None,
  None,
  "mumbai"
])
print(cities.value_counts())
print(cities.isnull())
print(cities.isnull().sum())

print(cities.notnull())
print(cities.notnull().sum())

#print(cities.dropna())
c = cities.fillna("Entertainment")
print(c)
```

```python
marks = pd.Series(["80", "90", "70"])

print(marks)
print(marks.dtype)

marks = marks.astype(int)

print(marks)
print(marks.dtype)
```
