# NumPy Array Copy vs View

```python
# COPY
arr = np.array([1, 2, 3, 4, 5])
x = arr.copy()  # the changes in x will not reflect in arr
arr[0] = 42

print(arr)
print(x)
```

```python
# VIEW:
arr = np.array([1, 2, 3, 4, 5])
x = arr.view() # the changes in x will reflect in arr 
arr[0] = 42

print(arr)
print(x)
```

# Check if Array Owns its Data
**As mentioned above, copies owns the data, and views does not own the data, but how can we check this?

Every NumPy array has the attribute base that returns None if the array owns the data.

Otherwise, the base  attribute refers to the original object.**

```python
arr = np.array([1, 2, 3, 4, 5])

x = arr.copy()
y = arr.view()

print(x.base)
print(y.base)
# The copy returns None.
# The view returns the original array.
```

# NumPy Joining Array
**Joining NumPy Arrays**

```python
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.concatenate((arr1, arr2))

print(arr)
```

```python
# Join two 2-D arrays along rows (axis=1):

arr1 = np.array([[1, 2], [3, 4]])

arr2 = np.array([[5, 6], [7, 8]])

arr = np.concatenate((arr1, arr2), axis=1) # concatenate means join

print(arr)
```

# Joining Arrays Using Stack Functions

```python
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.stack((arr1, arr2), axis=1) # axis can be changed 0 or 1 0=row, 1=column  # stack means join

print(arr)
```

# Stacking Along Rows
**NumPy provides a helper function: hstack() to stack along rows.**

```python
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.hstack((arr1, arr2))  # hstack means horizontal stack

print(arr)
```

# Stacking Along Columns
**NumPy provides a helper function: vstack()  to stack along columns.**

```python
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.vstack((arr1, arr2)) # vstack means vertical stack

print(arr)
```

# Stacking Along Height (depth)
**NumPy provides a helper function: dstack() to stack along height, which is the same as depth.**

```python
arr1 = np.array([1, 2, 3])

arr2 = np.array([4, 5, 6])

arr = np.dstack((arr1, arr2))

print(arr)
```

# NumPy Searching Arrays
You can search an array for a certain value, and return the indexes that get a match.

To search an array, use the where() method.

```python
arr = np.array([1, 2, 3, 4, 5, 4, 4])

x = np.where(arr == 4)  # it gives index number where values is 4

print(x)
```

```python
# Find the indexes where the values are odd:


arr = np.array([10, 14, 93, 41, 8, 7])

x = np.where(arr%2 == 1)

print(x)
```

```python
arr = np.array([10, 14, 93, 41, 8, 7])

x = np.where(arr%2 == 0)

print(x)
```

# Creating the Filter Array

```python
arr = np.array([41, 42, 43, 44])

# Create an empty list
filter_arr = []

# go through each element in arr
for element in arr:
  # if the element is higher than 42, set the value to True, otherwise False:
  if element > 42:
    filter_arr.append(True)
  else:
    filter_arr.append(False)

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])

# Create an empty list
filter_arr = []

# go through each element in arr
for element in arr:
  # if the element is completely divisble by 2, set the value to True, otherwise False
  if element % 2 == 0:
    filter_arr.append(True)
  else:
    filter_arr.append(False)

newarr = arr[filter_arr]

print(filter_arr)
print(newarr)
```

# Creating Filter Directly From Array
