## 3️⃣ Creating Arrays
NumPy provides multiple ways to create arrays:
- `np.array(): This is the most fundamental way to create an array. It converts a Python list, tuple, or other array-like object into a NumPy array.`
```
    import numpy as np
    my_list = [1, 2, 3, 4, 5]
    numpy_array = np.array(my_list)
    print(numpy_array)
```
- `np.arange(): This function creates an array with evenly spaced values within a given range, similar to Python's built-in range() function but returning a NumPy array. You can specify a start, stop, and step size.`
```
    import numpy as np
    arr_range = np.arange(0, 10, 2) # Array from 0 (inclusive) to 10 (exclusive), with a step of 2
    print(arr_range)
```
- `np.linspace(): This function generates an array with a specified number of evenly spaced values over a specified interval. You provide the start, stop, and the desired number of elements. `
```
    import numpy as np
    arr_linspace = np.linspace(0, 10, 5) # 5 evenly spaced values between 0 and 10 (inclusive)
    print(arr_linspace)
```
- `np.zeros(): This function creates an array of a given shape and data type, filled with zeros.`
```
    import numpy as np
    zeros_array = np.zeros((2, 3)) # A 2x3 array filled with zeros
    print(zeros_array)
```
- `np.ones(): Similar to np.zeros(), this function creates an array of a given shape and data type, filled with ones.`
```
    import numpy as np
    ones_array = np.ones((3, 2)) # A 3x2 array filled with ones
    print(ones_array)
```
- `np.eye(): This function creates a 2-D array with ones on the diagonal and zeros elsewhere, also known as an identity matrix. You specify the number of rows and columns (which are typically the same for an identity matrix).`
```
    import numpy as np
    identity_matrix = np.eye(3) # A 3x3 identity matrix
    print(identity_matrix)
```

````python
# 1D Array
a = np.array([1,2,3,4,5])
print(a)

# 2D Array
b = np.array([[1,2,3],[4,5,6]])
print(b)

# Specific dtype
c = np.array([1,2,3], dtype=float)
print(c)
````

**(b) Using `np.arange(start, stop, step)`**

````python
arr1=np.arange(0, 10)
print("arr1:", arr1)

arr2=np.arange(0, 20, 2)
print("arr2:", arr2)

arr3=np.arange(0, 1, 0.1)
print("arr3:", arr3)
````

**(c) Using `np.linspace(start, stop, num)`**

in this the logic is that you give the start and end num and num decides how many elements you want

````python
lin1 = np.linspace(0, 50, 5)
print("lin1:", lin1)

lin2 = np.linspace(10, 50, 9)
print("lin2:", lin2)
````

**(d) Using `np.zeros(shape)`**

````python
z1 = np.zeros(5) #1D array
print("z1:", z1)

z2 = np.zeros((2,3)) #2D array
print("z2:", z2)

z3 = np.zeros((3,3), dtype=int) #3D array
print("z3:", z3)
````

**(e) Using `np.ones(shape)`**

````python
o1 = np.ones(4) #1D array
print("o1:", o1)

o2 = np.ones((3,2)) #3D array
print("o2:", o2)

o3 = np.ones((2,2), dtype=int) #2D array
print("o3:", o3)
````

**(f) Using `np.eye(N)`** --> identity matrix

````python
I = np.eye(5)
print(I)

I2 = np.identity(3)
print("I2:", I2)
````
