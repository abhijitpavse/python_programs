```python
data = {
    "courses" : ["Python","Java","Data Science","Web development"],
    "students" : [45,35,60,50]
}
df = pd.DataFrame(data)

sns.set_theme()

sns.barplot(data=df,x="courses",y="students")

plt.title("Students enrolled in different courses")
plt.xlabel("Courses")
plt.ylabel("Number of Students")
plt.show()
```

# Difference between countplot and barplot is that countplot is used for count data and barplot is used for categorical data

```python
# distribtion: it tells us how value is spread accross dataset.
marks = [45,52,55,58,60,62,65,67,70,72,75,78,80,82,85,88,90,92,95,98]
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
sns.histplot(marks)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(marks,bins = 10)
plt.title("Distribution of student marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(marks,binwidth=5)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(marks,binwidth = 20)
plt.title("Distribution of student marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(marks,bins=6,binrange=(40,120),kde=True)
plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(marks,bins = 6, binrange = (40,120))
plt.title("Distribution of student marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.histplot(marks,bins = 6, binrange = (40,120), kde = True)
plt.title("Distribution of student marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
sns.kdeplot(marks, bw_adjust = 0.5)
plt.show()
```

```python
sns.rugplot(marks)
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.displot(marks,kind="kde")
plt.show()
```

```python
data = {
    "Category": [
        "Technology",
        "Furniture",
        "Office Supplies",
        "Technology",
        "Furniture",
        "Office Supplies",
        "Technology",
        "Furniture"
    ],
    "Sales": [
        1200,
        800,
        500,
        1500,
        900,
        600,
        1800,
        700
    ],
    "Profit": [
        300,
        100,
        80,
        450,
        150,
        120,
        500,
        90
    ],
    "Region":[
        "West",
        "East",
        "East",
        "South",
        "East",
        "North",
        "North",
        "South"
    ]
}

df = pd.DataFrame(data)

print(df)
```

```python
sns.countplot(data = df, x = "Category")
plt.title("Number of products by category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
sns.barplot(data = df, x = "Category", y = "Sales", estimator = "sum")
plt.title("Average Sales by category")

plt.show()
```

```python
sns.barplot(data = df, x = "Category", y = "Sales")
plt.title("median Sales by category")

plt.show()
```

```python
sns.boxplot(data = df, x = "Category", y = "Sales")
plt.title("Sale Distribution by category")

plt.show()
```

```python
sns.boxplot(data = df, x = "Category", y = "Sales", hue = "Region")
plt.title("Sale Distribution by category")

plt.show()
```

```python
sns.violinplot(data = df, x = "Category", y = "Sales")
plt.title("Sale Distribution by Category")

plt.show()
```

```python
sns.stripplot(data = df, x = "Category", y = "Sales")
plt.title("Sale Distribution by category")

plt.show()
```

```python
sns.swarmplot(data = df, x = "Category", y = "Sales")
plt.title("Sale Distribution by category")

plt.show()
```

```python
sns.swarmplot(data = df, x = "Category", y = "Sales", hue = "Region")
plt.title("Sale Distribution by category")

plt.show()
```

```python
sns.catplot(data = df, x = "Category", y = "Sales", hue = "Region", kind="bar")  #kinds = ["strip","swarm","box","boxen","violin","point","bar", "count"]
plt.title("Sale Distribution by Category")

plt.show()
```
